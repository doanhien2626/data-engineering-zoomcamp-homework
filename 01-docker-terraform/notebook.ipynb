```python
import pandas as pd

df = pd.read_parquet("green_tripdata_2025-11.parquet")
print(df.head())
```

```python
from sqlalchemy import create_engine
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
df.columns = df.columns.str.lower()
print(pd.io.sql.get_schema(df, name='green_tripdata_2025_11', con=engine))
```

```python
df.head(n=0).to_sql(name='green_tripdata_2025_11', con=engine, if_exists='replace')
```

```python
df.to_sql(
    name="green_tripdata_2025_11",
    con=engine,
    if_exists="append"
)
```
